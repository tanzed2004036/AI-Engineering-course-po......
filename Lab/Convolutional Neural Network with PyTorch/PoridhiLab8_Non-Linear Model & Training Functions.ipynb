Why Non-Linearity Matters

In Lab 01, our baseline model used only linear layers. The problem with stacking linear layers is that multiple linear transformations can always be reduced to a single linear transformation:

```
y = W2(W1(x)) = (W2 * W1)(x) = W_combined(x)
```

This means our "two-layer" network is mathematically equivalent to a single-layer network! No matter how many linear layers we stack, the model can only learn linear relationships and create straight-line decision boundaries.

The Solution: Activation Functions

By adding non-linear activation functions between layers, we break this linearity and allow the network to learn complex, non-linear patterns. The most popular activation function is

ReLU (Rectified Linear Unit)

.

The ReLU Activation Function

ReLU is simple yet powerful:

```
ReLU(x) = max(0, x)
```

If the input is

positive

→ output equals the input

If the input is

negative or zero

→ output is 0

![image](https://raw.githubusercontent.com/poridhiEng/lab-asset/8104ff41aaf569aa65977e43cdbadc13fc1b7a34/tensorcode/Deep-learning-with-pytorch/Computer-Vision/Lab_02/images/infra-4.png)

The graph shows the ReLU function with a "bend" at zero. The left side outputs 0 for all negative inputs, while the right side passes positive inputs through unchanged. This simple bend is what introduces non-linearity into our neural network!

Why ReLU?

Computational efficiency

- It's just a max operation, very fast to compute.

Solves vanishing gradient

- Gradients don't shrink for positive values.

Sparsity

- Many neurons output zero, making the network efficient.

Proven effectiveness

- Works well in practice across many domains.

Reusable Training Functions

In Lab 01, we wrote the training loop inline. As we build more models, this becomes repetitive. Let's create reusable functions:

train_step()

Performs one epoch of training by looping through all batches:

```
def train_step(model, data_loader, loss_fn, optimizer, accuracy_fn):
    model.train()
    for batch, (X, y) in enumerate(data_loader):
        y_pred = model(X)
        loss = loss_fn(y_pred, y)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
```

test_step()

Evaluates model on test data without computing gradients:
