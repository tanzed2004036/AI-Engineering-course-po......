What is Computer Vision?

Computer vision is the field of AI that enables computers to interpret and understand visual information. When we look at an image, we instantly recognize objects, people, and scenes. Teaching computers to do the same is the goal of computer vision.

![image](https://raw.githubusercontent.com/poridhiEng/lab-asset/8104ff41aaf569aa65977e43cdbadc13fc1b7a34/tensorcode/Deep-learning-with-pytorch/Computer-Vision/Lab_01/images/infra-4.svg)

Common Computer Vision Tasks

Binary Classification

is the simplest form where the model decides between two classes, answering yes/no questions like "Is this a cat or dog?"

Multiclass Classification

extends this to multiple categories, where the model must identify which single class an image belongs to from many options - for example, determining what animal is shown (cat, dog, lion, etc.).

Object Detection

goes further by not only classifying objects but also locating them within an image, such as finding all cars and pedestrians in a street photo.

Image Segmentation

provides the finest level of detail by classifying every pixel in an image, enabling tasks like separating foreground from background.

In this lab, we focus on

Multiclass Classification

- given an image of clothing, predict which of 10 categories it belongs to.

### 1️⃣ `ToTensor()` কী করে?

PyTorch-এর `ToTensor()` transform দুইটা কাজ করে:

* Image-এর pixel value সাধারণত থাকে:

**Step 2: DataLoader Batching**::😠❤️

Instead of processing one image at a time, we batch images together using DataLoader(dataset, batch_size=32, shuffle=True). The batch_size=32 means we process 32 images simultaneously, while shuffle=True randomizes the order each epoch to prevent the model from memorizing the sequence.

A batch of 32 images has shape [32, 1, 28, 28] following the NCHW format:



*   N = Number of images (32)
* C = Channels (1 for grayscale)
* H = Height (28)
* W = Width (28)

**Step 3: Model Architecture**

The diagram above shows how our baseline model processes image data. A batch of 28x28 grayscale images is first flattened into a 1D vector of 784 values (Input layer). This vector then passes through two linear layers: the first transforms 784 inputs to 10 hidden units, and the second produces 10 output values - one for each clothing category. Note that this baseline model uses no activation functions between layers.

Our baseline model processes data as follows:

* Layer	Input Shape	Output Shape	Description
* Input	[1, 28, 28]	-	28×28 grayscale image
* Flatten	[1, 28, 28]	[784]	Convert 2D to 1D vector
* Linear 1	[784]	[10]	First transformation (784→10)
* Linear 2	[10]	[10]	Output layer (10→10)
* Output	-	[10]	Raw scores (logits) for each class
Note: This baseline uses NO activation functions between layers. This is intentional - we want to establish a simple baseline first.

🛑 ***Step 5: Loss Function (CrossEntropyLoss)***

The loss function measures how wrong our predictions are:
CrossEntropyLoss:

* Takes raw logits (no need for softmax)

* Internally converts to probabilities

* Compares predicted probabilities to true labels

* Returns a single number (lower = better predictions)


````python
loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(predictions, true_labels)
````

Step 6: Optimizer (SGD)

The optimizer updates model weights to minimize loss:

````python
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
````

Stochastic Gradient Descent (SGD)

:

Uses gradients to determine which direction to adjust weights

Learning rate (lr=0.1)

: Controls step size

Too high → unstable training

Too low → very slow training

The Training Loop

Training repeats this cycle for each batch of data:

![image](https://raw.githubusercontent.com/poridhiEng/lab-asset/8104ff41aaf569aa65977e43cdbadc13fc1b7a34/tensorcode/Deep-learning-with-pytorch/Computer-Vision/Lab_01/images/infra-2.svg)

The 5 Steps (Per Batch)

````python
for epoch in range(epochs):
    for batch, (X, y) in enumerate(train_dataloader):

        # 1. Zero gradients (clear from previous step)
        optimizer.zero_grad()

        # 2. Forward pass (get predictions)
        y_pred = model(X)

        # 3. Calculate loss (how wrong are we?)
        loss = loss_fn(y_pred, y)

        # 4. Backward pass (compute gradients)
        loss.backward()

        # 5. Optimizer step (update weights)
        optimizer.step()
````

Step

Function

Purpose

1. Zero Gradients

optimizer.zero_grad()

Clear accumulated gradients from previous iteration

2. Forward Pass

model(X)

Pass data through model to get predictions

3. Compute Loss

loss_fn(pred, target)

Calculate how wrong predictions are

4. Backward Pass

loss.backward()

Compute gradients (how to adjust weights)

5. Optimizer Step

optimizer.step()

Update weights based on gradients

What is an Epoch?

1 Epoch

= One complete pass through the entire training dataset

With 60,000 images and batch_size=32: ~1,875 batches per epoch

We train for

3 epochs

= 5,625 total weight updates
